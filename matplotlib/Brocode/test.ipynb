```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("pokemon.csv")
freq = np.array(df['Type 1'].value_counts(sort=False))
names = list(df["Type 1"].unique())

figure, axes = plt.subplots(2,2)
plt.xticks(rotation=90, ha="right")

axes[0,0].plot(names, freq)
axes[0,0].tick_params(axis='x', labelrotation=90)

axes[0,1].scatter(names, freq)
axes[0,1].tick_params(axis='x', labelrotation=90)

axes[1,0].bar(names, freq)
axes[1,0].tick_params(axis='x', labelrotation=90)

axes[1,1].pie(freq, labels=names, radius = 1.5)

plt.tight_layout()
plt.show()
```
