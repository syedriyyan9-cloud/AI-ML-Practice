### Scatter Graph
shows the relationship between two variables.
helps to identify a correlation (+, -, None)
example: Study hours vs. test scores

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
study_hours = np.array([0,1,1,2,3,4,5,6,7,7,8,9])
test_scores = np.array([55,60,65,62,68,70,75,78,82,85,87,92])
plt.scatter(study_hours, test_scores, color="red", alpha=0.5, s=200, label="Class A")
plt.title("Study hours vs test scores")
plt.xlabel("Study Hours")
plt.ylabel("Test Scores in %")
plt.legend()
plt.show()
```
