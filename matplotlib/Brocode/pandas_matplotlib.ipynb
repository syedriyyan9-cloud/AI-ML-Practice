### Pandas + Matplotlib

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("pokemon.csv")
# print(df)
type2 = df['Type 2'].dropna()
# freq = np.array(type2.value_counts(sort=False))
type_count = type2.value_counts(ascending=True)
# print(freq, type(freq))
# types = list(type2.unique())
# print(types)
# print(len(freq) == len(types))
# plt.bar(types, freq)
plt.barh(type_count.index, type_count)
# plt.xticks(rotation=90)
plt.show()
```
