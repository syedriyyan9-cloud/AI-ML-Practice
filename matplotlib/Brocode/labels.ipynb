### labels

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# students in class
year = np.array([2023,2024,2025,2026])
c1 = np.array([15, 20, 30, 25])
c2 = np.array([45, 60, 15, 30])
c3 = np.array([30,35,40,41])

# styling
styling = dict(marker=".", markersize=10, linestyle='dashed')
title = dict(fontsize=10, family="Arial", fontweight="bold", color="black")

# labels
plt.title("Number of Students per Year in Each Class", **title)
plt.xlabel("Years", **title)
plt.ylabel("Students", **title)
plt.tick_params(axis="both", colors="brown")

# plot
plt.plot(year, c1, color="red", **styling)
plt.plot(year, c2, color="blue", **styling)
plt.plot(year, c3, color="green", **styling)
plt.xticks(year)
plt.show()
```
