### Pie Chart
pie chart is a circular chart divided into slices to show percentages of the total.
good for visualizing distribution among categories.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
fruits = ["Apples", "Bananas", "Cheeries", "Dragon Fruits"]
prices = np.array([300, 200, 150, 250])
colors = ["red", "yellow", "pink", "brown"]
plt.pie(prices, labels=fruits, autopct="%1.1f%%", colors=colors, explode=[0.1,0.1,0.1,0.1], shadow=True, startangle=90)
# plt.pie(prices, labels=fruits, autopct=lambda pct: f"${pct * 900 / 100:,.2f}", colors=colors, explode=[0.1,0.1,0.1,0.1], shadow=True, startangle=90)
plt.title("Prices of Fruits")
plt.show()
```
