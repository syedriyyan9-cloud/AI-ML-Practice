### grid() = helps make plots easier to read by adding reference lines.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
year = np.array([2023,2024,2025,2026])
c1 = np.array([15, 20, 30, 25])
c2 = np.array([45, 60, 15, 30])
c3 = np.array([30,35,40,41])

marker_style = dict(markersize = 10, marker="o",linestyle="solid")
title_style = dict(fontsize = 20, family="Arial", color="black")

plt.title("Number of student per year in a class", **title_style)
plt.tick_params(axis="both", colors = "blue")
plt.xlabel("Year", **title_style)
plt.ylabel("Students", **title_style)

# grid
# plt.grid(axis="both") # same as plt.grid(), nor args passed
# plt.grid(axis="x")
plt.grid(axis="y", linewidth=2, color="black", linestyle="dotted")

plt.plot(year, c1, color="red", **marker_style)
plt.plot(year, c2, color="green", **marker_style)
plt.plot(year, c3, color="blue", **marker_style)
plt.xticks(year)
plt.show()
```
