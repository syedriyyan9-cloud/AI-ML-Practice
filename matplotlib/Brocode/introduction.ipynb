### Introduction
Matplotlib is a python library, that is used for creating charts and graphs.
you can take plain numbers and turn them into visual you actually understand.
with just a few lines of code you can make a scrape

```python
import matplotlib.pyplot as plt
```

```python
plt.plot([1,2,3],[4,5,7])
plt.show()
```

you can also create line charts, bar charts, scatter graphs, histograms and pie charts.
you can also customize the appearence of your charts as well.


inside matplotlib there is a module called 'pyplot'.
pyplot provides a user friendly interface for plotting.

```python
# creating a basic plot
x = [2023, 2024, 2025, 2026]
y = [15, 25, 30, 20]

# the first arg is for the coordinates on the x-axis
# the second arg is for the coordinates on the y-axis
plt.plot(x, y)

# call show method to display plot
plt.show()

# passing only 1 arg to plot method would plot coordinates on y axis
```

```python
import numpy as np

# numpy arrays are faster than python lists
x = np.array([2023, 2024, 2025, 2026])
y = np.array([15, 25, 30, 20])
plt.plot(x, y)
plt.show()
```
