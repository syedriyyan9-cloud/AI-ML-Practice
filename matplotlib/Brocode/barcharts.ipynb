### bar chart
Bar chart compares categories of data by representing each category with a bar

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
categories = np.array(["Grains", "Fruits", "Vegetables", "Protien", "Dairy","Sweets"])
values = np.array([4, 3, 2, 5, 3, 1])
plt.title("Daily Consumption", color="red", fontfamily="Arial", size=15)
plt.xlabel("Food Categories", color="brown", fontfamily="calibri", size=20)
plt.ylabel("Amount", color="brown", fontfamily="calibri", size=20)
plt.grid(axis="y", linestyle="dotted", color='black', linewidth=1)
plt.bar(categories, values, color="skyblue")
# plt.barh(categories, values, color="blue")
plt.show()

```
