### Customization
this means to alter specifics of a graph according to your needs.

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.array([2023, 2024, 2025, 2026])
y = np.array([15, 25, 30, 20])

# marker takes the following arguments:
# ".", "o", "v", "*", "<", ">", "p","P","h","H","x" and many more
plt.plot(x, y, 
        marker=".", markersize=30, markerfacecolor="cyan",markeredgecolor = "red",
        linestyle="dashdot", linewidth=4, color="green")
plt.show()
```

```python
x = np.array([2023, 2024, 2025, 2026])
y1 = np.array([15, 25, 30, 20])
y2 = np.array([17, 23, 38, 5])
y3 = np.array([13,15,20,30])

line_style = dict(marker=".", 
                  markersize=30, 
                  markerfacecolor="cyan",
                  markeredgecolor = "red",
                  linestyle="dashdot", 
                  linewidth=4,) 
                #   color="green")

plt.plot(x, y1, color="red" , **line_style)
plt.plot(x, y2, color="green", **line_style)
plt.plot(x, y3, color="blue", **line_style)
plt.show()
```
