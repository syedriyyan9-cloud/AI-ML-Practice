### Array Attributes

#### basic Attributes of np array

```python
import numpy as np
import math
```

```python
a = np.arange(12).reshape(3,4)
print(a.shape)  # shape of matrix
print(a.ndim)   # dimensions of matrix
print(a.dtype)  # data type present in matrix
print(a.size)   # number of elements present in a matrix

print(len(a.shape) == a.ndim)           # shape gives len, ndim gives dimension
print(math.prod(a.shape) == a.size)     # product of tuple == number of elements, a.size also give number of elements  

print(type(a))
```
