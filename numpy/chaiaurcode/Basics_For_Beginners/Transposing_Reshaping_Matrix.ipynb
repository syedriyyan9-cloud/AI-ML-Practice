### Reshaping a Matrix

```python
import numpy as np
```

```python
a = np.array([[1,2,3], [4,5,6]])
print(a)
# reshaping array from a 2x3 matrix into a 3x2 matrix
b = a.reshape(3,2)
print(b)
```

### Transpose of a matrix

```python
# two ways to get the transponse of a matrix
# first is to use the transpose() method
# second is to use .T property

t = a.transpose()
print(t)
print("-------------")
t1 = a.T
print(t1)

```
