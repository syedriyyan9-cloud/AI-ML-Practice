### Generating Random Numbers

```python
import numpy as np
```

```python
rng = np.random.default_rng()
random = rng.random((3,2))
print(random)

random = rng.integers(5, size=(3,3))
print(random)
```

### Unique Items and Count

```python
# You can find the unique elements in an array easily with np.unique
a = np.array([11,11,12,13,14,15,16,17,12,13,11,14,18,19,20])
unique_value = np.unique(a)
print(unique_value)

# To get the indices of unique values in a NumPy array 
# (an array of first index positions of unique values in the array), 
# just pass the return_index argument in np.unique() as well as your array
unique_value, indices = np.unique(a, return_index = True)
print(indices)
```

```python
# You can pass the return_counts argument in np.unique() along with your array to get the frequency 
# count of unique values in a NumPy array.
freq = np.unique(a, return_counts = True)
print(freq)
```

```python
# this works on 2D arryas as well
a_2d = rng.integers(13, size=(4,4))
print(a_2d)
unique, indices, freq = np.unique(a_2d, return_index = True, return_counts=True)
print(unique)
print(indices)
print(freq)

# if no axis arg is given the a flattened array is returned

# using axis
unique = np.unique(a_2d, axis = 0)
print(unique)
```
