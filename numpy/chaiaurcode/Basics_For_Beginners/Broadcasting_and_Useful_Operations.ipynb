### Broadcasting

##### Broadcasting is a mechanism that allows NumPy to perform operations on arrays of different shapes. The dimensions of your array must be compatible

```python
import numpy as np
```

```python
a = np.array([2,2,2])
print(a * 3)
```

### Useful Array Operations

##### NumPy also performs aggregation functions. In addition to min, max, and sum, you can easily run mean to get the average, prod to get the result of multiplying the elements together, std to get the standard deviation, and more

```python
a = np.int64(np.random.rand(4) * 10)
print(a)
print(a.sum())
print(a.min())
print(a.max())
print(a.prod())
print(a.cumsum())
print(a.mean())
print(a.std())
```
