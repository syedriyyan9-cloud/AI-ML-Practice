### Creating an array from existing data using slicing, indexing, vstack, hstack, hsplit, view and copy

```python
import numpy as np
```

#### slicing

```python
a = np.arange(1,11)
a1 = a[2:5]
print(a1)
```

##### Vstack

```python
b = np.arange(1,5).reshape(2,2)
b1 = np.arange(6,10).reshape(2,2)
print(b)
print("------------")
print(b1)
print("------------")

v_stack = np.vstack((b,b1))
print(v_stack)
```

##### Hstack

```python
c = np.arange(1,5).reshape(2,2)
print(c)
print("------------")
c1 = np.arange(6,10).reshape(2,2)
print(c1)
print("------------")
h_stack = np.hstack((c,c1))
print(h_stack)
```

##### Hsplit

```python
d = np.arange(1,25).reshape(2,12)
print(d)
d1 = np.hsplit(d,3)
print(d1)
# If you wanted to split your array after the third and fourth column
d2 = np.hsplit(d, (3,4))
print(d2)
```

##### View
Views are an important NumPy concept! NumPy functions, as well as operations like indexing and slicing, will return views whenever possible.
This saves memory and is faster (no copy of the data has to be made). However it’s important to be aware of this - modifying data in a view also modifies the original array


```python
e = np.arange(1,13).reshape(3,4)
e1 = e[0,:]
print(e1)
e1[0] = 2342
print(e1)
print(e)
```

##### Copy
Using the copy method will make a complete copy of the array and its data (a deep copy).

```python
f = e1.copy()
print(f)
f[0] = 123123123
print(e1)
print(f)
```
