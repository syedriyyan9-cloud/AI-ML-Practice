### Create Basic Array

#### creating arrays using methods

```python
import numpy as np
```

```python
z = np.zeros(2)
print(z)
print(z.dtype)
print("--------------")
o = np.ones(2)
print(o)
print(o.dtype)
print("--------------")
```

#### The function empty creates an array whose initial content is random and depends on the state of the memory. The reason to use empty over zeros (or something similar) is speed - just make sure to fill every element afterwards

```python
e = np.empty(3)
print(e)
print(e.dtype)
print("----------")
ar = np.arange(4)
print(ar)
print(ar.dtype)
print("----------")
```

#### You can also use np.linspace() to create an array with values that are spaced linearly in a specified interval

```python
l = np.linspace(0,10,num=5)
print(l)
print(l.dtype)
print(len(l))

o1 = np.ones(2, dtype=np.int64)
print(o1.dtype)
```
