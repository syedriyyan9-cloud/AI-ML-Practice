### Reversing An Array

```python
import numpy as np
```

##### Reversing a 1D array

```python
arr = np.array([1,2,3,4,5])
print(arr)
rev_arr = np.flip(arr)
print(rev_arr)
```

##### Reversing a 2D array

```python
arr_2d = np.array([[1,2,3,4], [5,6,7,8]])
print(arr_2d)
rev_2d = np.flip(arr_2d)
print(rev_2d)

# to only reverse rows use axis as 0
rev_row = np.flip(arr_2d, axis=0)
print(rev_row)

# to only flip columns use axis as 1
rev_col = np.flip(arr_2d, axis=1)
print(rev_col)

# you can index a row and then reverse it
first_row = np.flip(arr_2d[0])
print(first_row)
```
