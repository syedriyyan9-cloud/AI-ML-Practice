### Indexing and Slicing

```python
import numpy as np
```

```python
a = np.arange(1,13).reshape(3,4)
print(a)

# in numpy we can pass expressions to filter out numbers
# print all even numbers in vector 'a'
print(a[a % 2 == 0])

# create a vector that contains values from 5 to 10
b = a[(a > 5) & (a <= 10)]
print(b)
# notice that we used tuples inside as only then would more then 1 expression is evaluated.

# You can also make use of the logical operators & and | in order to return boolean values that
# specify whether or not the values in an array fulfill a certain condition. 
# This can be useful with arrays that contain names or other categorical values
c = (a < 5) | (a > 11) # here we can see that tuples are not inside anything so we get answer in bool
print(c)

```

##### You can use np.nonzero() to print the indices of elements that are, for example, less than 5

```python
b = np.nonzero(a < 5)
print(b)

# a tuple of arrays was returned: one for each dimension. 
# The first array represents the row indices where these values are found, and 
# the second array represents the column indices where the values are found.

# If the element you’re looking for doesn’t exist in the array, 
# then the returned array of indices will be empty
print(np.nonzero(a == 100))
```
