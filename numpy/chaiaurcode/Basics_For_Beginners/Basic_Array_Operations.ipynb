### Basic Array operations

#### Adding arrays

```python
import numpy as np
```

```python
# Addition
data = np.array([4,4,4,4])
ones = np.ones(4, dtype=np.int64)
random = np.int64(np.random.rand((4))* 10)
print(random)
print(data + ones + random)
```

#### Subtracting Arrays

```python
# subtration
print(data - ones - random)
```

#### Multipling Arrays

```python
# Multiplication
print(data * ones * random)
```

#### Dividing Arrays

```python
# division
# '/' gives answer in float
# '//' gives answer in int
print(data // ones)
```

#### If you want to find the sum of the elements in an array, you’d use sum()

```python
print(data.sum())
print(ones.sum())
print(random.sum())
```

#### To add the rows or the columns in a 2D array, you would specify the axis.

```python
b = np.array([[1,1],[2,2]])
print(b.sum(axis=0)) # row wise addition -> 1 + 2, 1 + 2 = [3, 3]
print(b.sum(axis=1)) # col wise addition -> 1 + 1, 2 + 2 = [2, 4]
```
