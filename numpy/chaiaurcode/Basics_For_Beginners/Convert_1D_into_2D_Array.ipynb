### Convert 1-Dimensional array into 2-Dimensional array

#### You can use np.newaxis and np.expand_dims to increase the dimensions of your existing array.

##### Using np.newaxis will increase the dimensions of your array by one dimension when used once. This means that a 1D array will become a 2D array, a 2D array will become a 3D array, and so on.

```python
import numpy as np
```

```python
a = np.arange(9)
print(a.shape)
```

```python
b = a[np.newaxis, :]
print(b.shape)
print(b)

c = a[np.newaxis, :]    # returns only modified array, does not modify the original array
print(c.shape)
print(c)

d = a[np.newaxis, :]    # increases a dimension in row
print(d.shape)
print(c)

d = a[:, np.newaxis]    # increases a dimension in column
print(d.shape)
print(d)
```

##### You can also expand an array by inserting a new axis at a specified position with np.expand_dims

```python
# the .shape method returns the tuple which tells use the dimensions of a vector
# however it can be indexed as it is a tuple
# so using .expand_dims method we can pass the vector name and the index at which we want to expand a dimension

a = np.arange(9)
print(a.shape)

b = np.expand_dims(a, axis=1)   # does not modify 'a' 
print(b.shape)
print(b)

c = np.expand_dims(b, axis=2)
print(c.shape)
print(c)
```
