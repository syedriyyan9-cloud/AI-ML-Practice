### Starting with basic operation on numpy arrays

```python
import numpy as np
```

##### Arithmetic operatiors on arrays apply elementwise

```python
a = np.array([10,20,30,40,50])
b = np.arange(1,6)
c = a - b
print(c)

print(b ** 2)

print(10 * np.sin(a))

print(a < 100)

# the arrays a and b remain unchanged
print(a, b)
```

##### the product using * operator operates element wise
##### the matrix product uses @ or .dot() method and multiplies operates row wise

```python
# a1 = np.ones((5), dtype=np.int64)
# a2 = (np.random.rand(5) * 10).astype(np.int64)
a1 = np.arange(4).reshape(2,2)
a2 = np.array([[2,0],[6,4]])
# print(a1, a2,sep="\n")
a3 = a1 * a2
# print(a3)
# print(np.dot(a1, a2))
print(a1@a2)
```

##### Some operations, such as += and *=, act in place to modify an existing array rather than create a new one.

```python
b1 = np.ones((2,2)).astype(np.int64)
b2 = (np.random.rand(4) * 10).reshape((2,2)).astype(np.float32)

b1 *= 3
print(b1)

print(b2)
b2 += 1
print(b2)

# before adding two matrix together ensure they are of same data type
b1 += b2
```

##### Many unary operations, such as computing the sum of all the elements in the array, are implemented as methods

```python
c1 = (np.random.rand(9) * 10).reshape(3,3).astype(np.int64)
print(c1)
print(c1.sum())
print(c1.max())
print(c1.min())

# specifying the axis parameter you can apply an operation along the specified axis of an array
print(c1.sum(axis=1))   # row
print(c1.min(axis=0))   # col
print(c1.max(axis=1))   # row
```
