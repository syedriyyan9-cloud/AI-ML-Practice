#### The ix_ function can be used to combine different vectors so as to obtain the result for each n-tuple. For example, if you want to compute all the a+b*c for all the triplets taken from each of the vectors a, b and c

```python
import numpy as np
```

```python
a = np.array([1,2,3])
b = np.array([4,5,6])
c = np.array([7,8,9])
ax, bx, cx = np.ix_(a,b,c)
print(ax)
print("---------")
print(bx)
print("---------")
print(cx)
print("---------")
result = ax + bx * cx
print(result)
```
