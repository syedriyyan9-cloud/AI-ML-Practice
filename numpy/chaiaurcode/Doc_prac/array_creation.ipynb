### understading Array creating in numpy

```python
import numpy as np
```

```python
a = np.arange(20).reshape(5,4) # create an array of 20 elements from 0 to 19 and then arrange (reshape) them into 5 rows and 4 columns
print(a)
print(a.ndim)   # how many dimensions (axes) does the array have
print(a.shape)  # what are the number of rows and columns in array
print(a.size)   # how many elements are there in the array
print(a.itemsize)   # how much memory space is one element taking
print(a.dtype)  # what is the data type of the elements
```

```python
b = np.array([[[1,2,3], [4,5,6]],[[3,2,1],[6,5,4]]]) # 3 dimensional array
print(b.ndim)
print(b.dtype) # int64
print(type(b))
c = np.array([1,2,3,4], dtype=np.float64)   # defining data type of array while creating it.
print(c.dtype)
```

##### The function zeros creates an array full of zeros
##### the function ones creates an array full of ones
##### the function empty creates an array whose initial content is random and depends on the state of the memory. 
##### By default, the dtype of the created array is float64, but it can be specified via the key word argument dtype.

```python
z = np.zeros((2,2))
o = np.ones((3,2))
r = np.empty((3,3))
print("this is array containing zeros")
print(z)
print("this is array containing ones")
print(o)
print("this is array containing random numbers")
print(r)
```

##### To create sequences of numbers, NumPy provides the arange function which is analogous to the Python built-in range, but returns an array.

```python
print(np.arange(10))
print(np.arange(10, 21, 2))
# also takes floats as arguments
print(np.arange(0,2,0.3))
```

##### using arange for generating floats is not a good idea as the sequence cannot be predictied due to finite floating point precision error. so for this reason use "linspace"

```python
print(np.linspace(0,2,9)) # have 9 numbers in an array from 0 to 2
```
