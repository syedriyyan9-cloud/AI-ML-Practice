### Learn about numpy with chai aur code

```python
import numpy as np
```

Creating array from list

```python
arr_1d = np.array([1,2,3,4,5])
print("1D array: ", arr_1d)
arr_2d = np.array([[1,2,3],[4,5,6]])
print("2D array: ", arr_2d)
```

### List vs numpy array

```python
py_list = [1,2,3]
print("python list multiplication: ", py_list * 2)

np_array = np.array([1,2,3])  #element wise multiplication
print("numpy list multiplication: ", np_array * 2)

import time
start = time.time()
py_list = [i * 2 for i in range(1000000)]
print("\n list operation time: ", time.time() - start)

start = time.time()
np_array = np.arange(1000000) * 2
print("\n numpy operation time: ", time.time() - start)
```

### creatint array from scratch

```python
zeros = np.zeros((3,4))
print("zeros array: \n", zeros)

ones = np.ones((4,4))
print("ones array: \n",ones)

full = np.full((2,2), 7)
print("full array: \n",full)

random = np.random.random((2,3))
print("random array: \n",random * 10)

sequence = np.arange(0, 10, 2)
print("sequence array: \n",sequence)
```

### Vector, Matrices and Tensor

```python
vector = np.array([1,2,3])
print("vector: \n", vector)

matrix = np.array([[1,2,3], [4,5,6]])
print("matrix: \n", matrix)

tensor = np.array([[[1,2], [3,4]],
                   [[5,6],[7,8]]])
print("Tensor: ", tensor)
```

### array properties

```python
arr = np.array([[1,2,3],
                [4,5,6]])
print("Shape ", arr.shape)
print("Dimension ", arr.ndim)
print("Size (elements present) ", arr.size)
print("Data types ", arr.dtype)
```

### array reshaping

```python
arr = np.arange(12)
print("original array: ", arr)

reshaped = arr.reshape((3,4))
print("Reshaped array: ", reshaped)

flattened = reshaped.flatten()
print("Flattened array: ", flattened)

# revel (returns view, instead of copy)
raveled = reshaped.ravel()
print("Raveled array: ", raveled)

# transpose
transpose = reshaped.T
print("Transposed array: ", transpose)

```
