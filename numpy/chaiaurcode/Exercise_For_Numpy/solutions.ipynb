#### 13. Create a 10x10 array with random values and find the minimum and maximum values (★☆☆) (Solution)

````python
random = np.int64(np.random.rand(100) * 1000).reshape(10,10)
print(random)
minimum = np.min(random)
maximum = np.max(random)
print(minimum, maximum)
````

#### 14. Create a random vector of size 30 and find the mean value (★☆☆) (Solution)

````python
random = np.int64(np.random.rand(30) * 100)
mean = random.mean()
print(mean)
````

#### 15. Create a 2d array with 1 on the border and 0 inside (★☆☆) (Solution)

````python
border = np.zeros(20).reshape(5,4)
border[:1,] = 1
border[:,0] = 1
border[:,-1] = 1
border[-1,] = 1
print(border)
````

#### 17. What is the result of the following expression? (★☆☆) (Solution)
```python
0 * np.nan
np.nan == np.nan
np.inf > np.nan
np.nan - np.nan
np.nan in set([np.nan])
0.3 == 3 * 0.1
```

#### 18. Create a 5x5 matrix with values 1,2,3,4 just below the diagonal (★☆☆) (solution)

````python
diagnol = np.zeros(25).reshape(5,5)
diagnol[1,0] = 1
diagnol[2,1] = 2
diagnol[3,2] = 3
diagnol[4,3] = 4
print(diagnol)
````

#### 19. Create a 8x8 matrix and fill it with a checkerboard pattern (★☆☆) (Solution)

````python
checkerboard = np.zeros(64).reshape(8,8)
checkerboard[1::2,::2] = 1
checkerboard[::2, 1::2] = 1
print(checkerboard)
````

#### 20. Consider a (6,7,8) shape array, what is the index (x,y,z) of the 100th element? (★☆☆) (Solution)

````python
arr = np.arange(6*7*8).reshape(6,7,8)
index = np.where(arr == 100)
print(index)
````
