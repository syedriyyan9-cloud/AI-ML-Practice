#### 21. Create a checkerboard 8x8 matrix using the tile function (★☆☆) (Solution)

````python
boxes = np.array([0,1])
checkerboard = np.tile(boxes, (8,4))
checkerboard[1::2,] = 1
checkerboard[1::2, 1::2] = 0
print(checkerboard)
````

#### 22. Normalize a 5x5 random matrix (★☆☆) (Solution)

````python
five_by_five = np.random.rand(25).reshape(5,5)
minn = np.min(five_by_five)
maxx = np.max(five_by_five)
norm = (five_by_five - minn) / (maxx - minn)
print(norm)
````

#### 33. How to get the dates of yesterday, today and tomorrow? (★☆☆) (Solution)

````python
today = np.datetime64('today', 'D')
yesterday = today - np.timedelta64(1, 'D')
tomorrow = today + np.timedelta64(1, 'D')
print(yesterday, today, tomorrow)
````

#### 35. How to compute ((A+B)*(-A/2)) in place (without copy)? (★★☆) (solution)

````python
A = np.array([1.,2.,3.])
B = np.array([4.,5.,6.])
np.add(A,B,out=B)
np.negative(A, out=A)
np.divide(A, 2, out=A)
np.multiply(A,B,out=A)
print(A)
````
