#### arrays can be indexed by arrays of integers and arrays of booleans

```python
import numpy as np
```

```python
a = np.arange(12) * 10
print(a)
print("------------")
i = np.array([5,6])
print(a[i])
```

#### When the indexed array is multidimensional, a single array of indices refers to the first dimension

```python
a = np.arange(12).reshape(3,4)
print(a)
print("------------")
i = np.array([[0,1,0],[0,2,0]]) # number of indices in both inner arrays should be same as the first dim of a
print(a[i])
```

#### We can also give indexes for more than one dimension. The arrays of indices for each dimension must have the same shape.

```python
b = np.arange(12).reshape(3,4)
print(b)
r = np.array([0,1,0]) # first dim indices, must have same length as c
c = np.array([0,1,3]) # second dim indices, must have same lenght as r
print(b[r,c]) # r is for rows, c is for columns

# in a diagnol we would have same values of indices for r, c
# example: r = [0,1,2] and c = [0,1,2]

print(b[:,2])

# In Python, arr[r, c] is exactly the same as arr[(r, c)]—so we can put r and c in a tuple and then do the indexing with that.
l = (r,c)
print(b[l])

# we can not do this by putting r and c into an array, because this array will be interpreted as indexing the first dimension of b.
s = np.array([r,c])
print(b[s]) # gives error
```

#### You can also use indexing with arrays as a target to assign to

```python
d = np.arange(5)
print(d)
d[[0,2,4]] = 1
print(d)

# However, when the list of indices contains repetitions, the assignment is done several times, leaving behind the last value:
print("----------")
d = np.arange(5)
d[[0,0,2]] = [1,2,3]
print(d)

# This is reasonable enough, but watch out if you want to use Python’s += construct, as it may not do what you expect:
print("----------")
d = np.arange(5)
d[[0,0,2]] += 1
print(d)
# Even though 0 occurs twice in the list of indices, the 0th element is only incremented once.
```
