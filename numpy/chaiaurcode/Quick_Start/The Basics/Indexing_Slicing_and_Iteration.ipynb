### One-dimensional arrays can be indexed, sliced and iterated over, much like lists and other Python sequences.


```python
import numpy as np
```

```python
a = np.arange(5)
print(a)

b = a[1:4]
print(b)

print(a[:]) # print all values in a
print(a[::2]) # print every even index of a
print(b[1:3] * 10)

# print table from 1 to 10 in a matrix
a = np.arange(1, 11)
b = np.arange(1,11).reshape(10,1)
print(a * b)
```

```python
# 2d array indexing
c = np.arange(6).reshape(2,3)
print(c)

# slicing on row
print(c[0]) # get first row
print(c[1]) # get second row

print(c[0,0]) # get first element of first row
print(c[1,0]) # get first element of second row
# 0, 1
print(c[0, 0:2])
# 3, 4
print(c[1, 0:2])

# get[[1,2],[4,5]]
print(c[:,1:3])
```

### The dots (...) represent as many colons as needed to produce a complete indexing tuple.

```python
three_dim = np.arange(8).reshape(2,2,2)
print(three_dim.ndim)
print(three_dim.shape)
print(three_dim)

# using dots(...)
# get first element of every array
print(three_dim[...,0])

# if one wants to perform an operation on each element in the array, one can use the flat attribute which is an iterator over all the elements of the array
for element in three_dim.flat:
    print(element)
```
