### Data Cleaning
data cleaning = the process of fixing/removing:
                incomplete, incorrect or irrelevant data.
                ~75% of work done with pandas is data cleaning

```python
import pandas as pd
file = "pokemon.csv"
df = pd.read_csv(file)
```

```python
# drop irrelevant columns
# .drop method returns a new modified dataframe
dropped = df.drop(columns=["Legendary"])
print(dropped)
```

```python
# handle missing data
# drop rows that are missing values
dropped = df.dropna(subset=['Type 2'])
print(dropped)
```

```python
# insead of dropping fill not available values
fill = df.fillna({"Type 2": 'None'})
print(fill)
```

```python
# fix inconsistent values
df["Type 2"] = df["Type 2"].replace({None: "GRASS"})
print(df.to_string())
```

```python
# standardize text
df['Name'] = df['Name'].str.lower()
print(df.to_string())
```

```python
# fix data types
df["Legendary"] = df["Legendary"].astype(int)
print(df)
```

```python
# remove duplicate values
dup = df.drop_duplicates()
print(dup)
```
