### DataFrame = A tabular data strcuture with rows and columns. (2 Dimensional)
### Similar to an Excel Spreadsheet

```python
import pandas as pd
```

```python
data = {
    "Name": ["syed", "riyyan", "hassan"],
    "Age": [12,23,34]
}
# dataframe short form df
df = pd.DataFrame(data, index=["Employee 1", "Employee 2", "Employee 3"])
print(df)

print(df.loc["Employee 1"])
print(df.iloc[1])

# add a new column
df["Job"] = ["Student", "Teacher", "N/A"]
print(df)

# add a new row
# to add multiple rows use more than one dicts
new_row = pd.DataFrame([{"Name": "Sandy", "Age": 28, "Job": "Engineer"}], index=["Employee 4"])
df = pd.concat([df,new_row])
print(df)
```
