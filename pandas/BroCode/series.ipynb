#### Series
A pandas 1 diemensional labeled array that can hold any data type. Thin of it like a single column in a spreadsheet (1 dimensional)

```python
data = [1,2,3]  # python list
# converting python list to pandas series
series = pd.Series(data)    # we call the constructor Series, hence 'S' is capital
print(series)

# different data types
strings = ["A","B","C"]
booleans = [True, False, True]
integers = [1,2,3]
floats = [1.3, 2.6, 3.9]

# converting all of the above data types to series
s1 = pd.Series(strings)
s2 = pd.Series(booleans)
s3 = pd.Series(integers)
s4 = pd.Series(floats)
print(s1)
print(s2)
print(s3)
print(s4)
```

```python
# we can change index using index argument in the constructor
data = [1,2,3]
series = pd.Series(data, index=['a', 'b', 'c'])
print(series)

# to select a value from series (column in a spreadsheet) use 'loc' method, 
# loc stands for 'location by label'
print(series.loc["a"])

# if label does not exist like 'd' then we get a key error
```

```python
# changing the value of a label
series.loc['c'] = 1000
print(series)

# we can also acccess the values by integer position using iloc method
print(series.iloc[1])   # works like array indexing
```

```python
# filter by values
data = [100, 102, 104, 200, 202]
series = pd.Series(data, index=['a', 'b', 'c', 'd', 'e'])
print(series[series >= 200])
print(series[series < 200])

# using python dict
calories = {'day 1': 1750, 'day 2': 2100, 'day 3': 1700}
series = pd.Series(calories)    # not passing an index here would give keys as labels
print(series)
series.loc["day 3"] += 500
print(series.loc['day 2'])
print(series.iloc[0])
print(series[series >= 2000])
```
