### Aggregate Functions
aggregate functions     =   reduces a set of vlaues into a single summary value
                            Used to summarize and analyze data
                            Often used with the groupby() function

Aggregation in pandas is the process of taking a collection of multiple data points and compressing them into a single, meaningful summary value. Aggregate functions are the mathematical operations (like calculating a sum, average, or maximum) used to perform this data reduction.

Methods             Operations

.sum()	            Calculates the total of all numeric values
.mean()	            Computes the arithmetic average
.median()           Finds the middle value of ordered observations
.min() / .max()	    Returns the smallest or largest value
.count()	        Counts non-null/non-missing entries
.nunique()	        Counts the number of distinct unique values
.std() / .var()	    Measures data volatility (standard deviation/variance)

```python
import pandas as pd
```

```python
file = "pokemon.csv"
df = pd.read_csv(file)
```

```python
# mean method cannot be used on non numeric columns
# numeric_only arg states to only apply on numeric columns
print(df.mean(numeric_only=True))

# sum method
print(df.sum(numeric_only=True))
```

```python
# max method
print(df.max(numeric_only=True))

# min method
print(df.min(numeric_only=True))
```

```python
# count method
print(df.count(numeric_only=True))
```

The above examples work on entire dataframe, but what if we want to only work on one column

```python
# syntax to apply aggregate functions on a specific column
# dataframe['col_name'].aggregate_method()
print(f"Mean: {df['HP'].mean()}")
print(f"Count: {df['HP'].count()}")
print(f"Max: {df['HP'].max()}")
print(f"Min: {df['HP'].min()}")
print(f"Median: {df['HP'].median()}")
print(f"Unique: {df['HP'].nunique()}")
```

.groupby(). This splits your dataset into categories before collapsing each category into a single metric.

```python
# grouping each pokemon by their type 2 abilities and then finding the mean HP of each group
group = df.groupby('Type 2')
print(group["HP"].mean())
```

```python
print(group["HP"].count())
```
