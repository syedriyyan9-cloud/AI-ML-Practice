### Filtering 
filtering = keeping the rows that match a condition

```python
import pandas as pd
```

```python
file = "pokemon.csv"
df = pd.read_csv(file)
print(df)
```

```python
# legendary_pokemon = df[df["Legendary"] == True][["Name", "HP", "Attack", "Defense"]].to_string()
legendary_pokemon = df[df["Legendary"] == True] # df[condition]
print(legendary_pokemon[["Name", "Attack", "Defense"]].to_string())

# filtering syntax:
# df[condition]
# condition = df['col_name' then comparision operator then number, string etc..]
# so filtering become df[condition] == df[df[actual condition]]
# to enable selection after it:
# df[condition][[select cols]]
```

```python
gp_pokemon = df[(df["Type 1"] == "Grass") & (df["Type 2"] == "Poison")]
print(gp_pokemon)
```

```python
# filter pokemons that are either fire and flyting type or water and poision type
f_and_f_or_w_and_p_pokemon = df[(df["Type 1"] == "Fire") & (df["Type 2"] == "Flying") | (df["Type 1"] == "Water") & (df["Type 2"] == "Poison")]
print(f_and_f_or_w_and_p_pokemon.to_string())
```

```python
# filter pokemons that have hp greater than 100 and attack greater than 110
hp_a = df[(df["HP"] > 100) & (df["Attack"] > 110)]
print(hp_a.to_string())
```
