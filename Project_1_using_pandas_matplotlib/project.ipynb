```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("salaries.csv")

# Show shape, dtypes, and info().
print(df.shape)
print(df.dtypes)
print(df.info())
```

```python
# Count missing values per column.
print(df.isna().sum())

# Describe salary_in_usd (count, mean, median, min, max).
print(df["salary_in_usd"].describe())

# What is the median salary overall?
print(df["salary_in_usd"].median())

# How many unique job_title values exist?
print(df["job_title"].unique())
```

```python
# Drop duplicates.
df.drop_duplicates()

# Convert work_year to int (if not already).
print(df["work_year"].dtypes)

# Verify no missing values in salary_in_usd.
print(df["salary_in_usd"].isna().sum())
```

```python
# Median salary by experience_level — which level earns most?
print(df.groupby("experience_level")["salary_in_usd"].max())

# Median salary by company_size.
print(df.groupby("company_size")["salary_in_usd"].median())

```

```python
# Median salary by job_title — top 10 highest-paying roles.
highest = df.groupby("job_title")["salary_in_usd"].median().sort_values()
highest_pay = highest[-10:]
highest_job = highest_pay.index
# print(highest)
print(highest_pay)
print(highest_job)
```

```python
# Median salary by remote_ratio — does remote work pay more?
highest = df.groupby("remote_ratio")["salary_in_usd"].median().sort_values()
print(highest)

# Add a salary_level column: "high" if salary_in_usd > overall median else "low".
median = df["salary_in_usd"].median()
df["salary_level"] = np.where(df["salary_in_usd"] > median, "high", "low")
print(df["salary_level"])

```

```python
# For each experience_level, what % of rows are "high" salary?
exp_level = sorted(df["experience_level"].unique())
exp_level_group = df.groupby("experience_level")["salary_level"].count()
high_rows = list()
for x in exp_level:
    high_rows.append((df[(df["salary_level"] == "high") & (df["experience_level"] == x)].count().loc["experience_level"]))
total_rows = list()
for x in exp_level_group:
    total_rows.append(x)
high_rows = np.array(high_rows)
total_rows = np.array(total_rows)  
print((high_rows / total_rows) * 100)  

```
